```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import math
```

```python
x = np.array([25836, 13628, 15396, 25792, 30704, 14816, 30548, 33460, 25632, 34144]) #Data for 250 simulations
y = np.array([26088, 26032, 15048, 15180, 14896, 21888, 25412, 25808, 30864, 14512]) #Data for 100 simulations
n = 10
df = n - 1

x_hat = x.mean()
y_hat = y.mean()

std_x = np.var(x)**.5
std_y = np.var(y)**.5

print(x_hat, std_x)
print(y_hat, std_y)

#Calculates 95% CI for x and y
t_95 = stats.t.ppf(q=0.95, df=df)

standard_error_of_mean_x = std_x / n**.5
standard_error_of_mean_y = std_y / n**.5

margin_of_error_x = t_95 * standard_error_of_mean_x
margin_of_error_y = t_95 * standard_error_of_mean_y

lower_limit_x = round(x_hat - margin_of_error_x, 2)
upper_limit_x = round(x_hat + margin_of_error_x, 2)

lower_limit_y = round(y_hat - margin_of_error_y, 2)
upper_limit_y = round(y_hat + margin_of_error_y, 2)

print(f"95 CI for 250 simulations: lower limit: {lower_limit_x} and upper limit: {upper_limit_x}")
print(f"95 CI for 100 simulations: lower limit: {lower_limit_y} and upper limit: {upper_limit_y}")
```

```python
#Calculating the sample size needed to have a 95% CI with a standard deviation of 1000.
#This i done by isolating n in the SE_mean formula where we set the SE_mean to 1000.
#n = (std / SE_mean)**2
se_mean = 1000
required_n_x = (std_x / se_mean)**2
required_n_y = (std_y / se_mean)**2
print(math.ceil(required_n_x))
print(math.ceil(required_n_y))
#Here it makes sense that we need a smaller required sample size for 100 simulations, since it has a smaller std. 
```
